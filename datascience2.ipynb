```python
import matplotlib.pyplot as plt
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]
plt.bar(x, y)
plt.show()

```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Bar Chart')
plt.bar(x, y)
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Bar Chart')
plt.bar(x, y, color='red', edgecolor='yellow', linewidth=2, alpha=0.9)
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]

plt.xlabel('X-axis', fontsize=14, color='red')
plt.ylabel('Y-axis', fontsize=14, color='blue')
plt.title('Bar graph', fontsize=16, color='cyan')

c = ["yellow", "green", "blue", "red", "cyan"]

plt.bar(x, y, color=c, width=2)

plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]

plt.xlabel('X-axis', fontsize=14, color='red')
plt.ylabel('Y-axis', fontsize=14, color='blue')
plt.title('Bar graph', fontsize=16, color='cyan')

c = ["yellow", "green", "blue", "red", "cyan"]

plt.bar(x, y, color=c, width=2, edgecolor='black', linewidth=2, alpha=0.8,linestyle='--', hatch='\\')

plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]

plt.xlabel('X-axis', fontsize=14, color='red')
plt.ylabel('Y-axis', fontsize=14, color='blue')
plt.title('Bar graph', fontsize=16, color='cyan')

c = ["yellow", "green", "blue", "red", "cyan"]

plt.bar(x, y, color=c, width=2, edgecolor='black', linewidth=2, alpha=0.8,linestyle='--', hatch='\\')
plt.legend("homo")
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Bar Chart')
plt.barh(x, y, color='red', edgecolor='yellow', linewidth=2, alpha=0.9)
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Bar Chart')
plt.scatter(x, y, color='red', edgecolor='yellow', linewidth=2, alpha=0.9)
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]

plt.xlabel('X-axis', fontsize=14, color='red')
plt.ylabel('Y-axis', fontsize=14, color='blue')
plt.title('Bar graph', fontsize=16, color='cyan')

c = ["yellow", "green", "blue", "red", "cyan"]
size = [100, 200, 300, 400, 500]

plt.scatter(x,y,color = c,alpha=0.9, marker="*", edgecolor = "black",s=size, linewidth = 2)

plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [100, 200, 300, 400, 500]

plt.xlabel('X-axis', fontsize=14, color='red')
plt.ylabel('Y-axis', fontsize=14, color='blue')
plt.title('Bar graph', fontsize=16, color='cyan')

c = ["yellow", "green", "blue", "red", "cyan"]##range of color between 1  to 100
size = [100, 200, 300, 400, 500]

plt.scatter(x,y,color = c,alpha=0.9, marker="o", edgecolor = "black",s=size, linewidth = 2)#alpha range(0-0.1)
t = plt.colorbar()
t.set_label('Colorbar', fontsize=14, color='green')

plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [2,4,1,5,8]
plt.pie(x,labels=y, explode=(0, 0, 0, 0.6, 0), shadow=True, startangle=90)
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [2,4,1,5,8]
plt.pie(x,labels=y, explode=(0, 0, 0, 0.6, 0), shadow=True, startangle=90,autopct='%0.1f%%')
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [2,4,1,5,8]
plt.pie(x,labels=y, explode=(0, 0, 0, 0.6, 0), shadow=True,radius=1.0, startangle=90,autopct='%0.1f%%')
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [2,4,1,5,8]
plt.pie(x,labels=y, explode=(0, 0, 0, 0.6, 0), shadow=True,radius=1.0, startangle=90,autopct='%0.1f%%',labeldistance=1.1)
plt.show()
```

```python
x = [10, 20, 30, 40, 50]
y = [2,4,1,5,8]
plt.pie(x,labels=y, explode=(0, 0, 0, 0.6, 0), shadow=True,radius=1.0, startangle=90,autopct='%0.1f%%',labeldistance=1.1,
        textprops={ "fontsize":15},counterclock = False, wedgeprops={"linewidth":2,"edgecolor":"yellow"} )
plt.title("pie plot")
plt.legend(loc=2)
plt.show()
```
