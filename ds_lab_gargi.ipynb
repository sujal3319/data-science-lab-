```python
##vectorised operation 
arr1=np.array([1,2,3,4,5])
arr2=np.array([10,20,30,40,50])

##elelment wise addition 
print("addition", arr1+arr2)
print("substraction", arr1-arr2)
print("multiplication", arr1*arr2)
print("division", arr1/arr2)
```

```python
##universal function 
arr = np.array([1,.2,3,4,5])

## square root 
print(np.sqrt(arr))
print(np.exp(arr))##exponential
print(np.sin(arr))##sine function
print(np.log(arr))##log function
```

```python
##noramlisation
data =np.array([1,2,3,4,5])
mean=np.mean(data)#MEAN
std_dev=np.std(data)#STANDARD DEVIATION 
normalised_data =(data-mean)/std_dev
print(mean)
print(std_dev)
print("normalised data:", normalised_data)
```

```python
#statistic concept
arr2 = np.array([1,2,3,45,5])
mean = np.mean(arr2)
print(mean)

median = np.median(arr2)
print(median)

variance =np.var(arr2)
print(variance)

std_dev =np.std(arr2)
print(std_dev)
```
